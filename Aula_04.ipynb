```python
def funcaoCalcularMedia(nota1, nota2):
  soma = nota1 + nota2
  media = soma / 2
  return media

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = funcaoCalcularMedia(nota1, nota2)
print("A média das notas é: {}".format(media))
```

```python
def is_prime(num):
    if num <= 1:
        return False
    if num == 2:
        return True
    if num % 2 == 0:
        return False

    i = 3
    while i * i <= num:
        if num % i == 0:
            return False
        i += 2
    return True

number = int(input("Digite um número para verificar se é primo: "))
if is_prime(number):
    print(f"{number} é um número primo.")
else:
    print(f"{number} não é um número primo.")
```

```python
def list_first_n_primes(n):
    if n <= 0:
        return []

    primes = []
    count = 0
    num = 2
    while count < n:
        if is_prime(num):
            primes.append(num)
            count += 1
        num += 1
    return primes


num_primes = int(input("Digite a quantidade de números primos iniciais para listar: "))
prime_list = list_first_n_primes(num_primes)
print(f"Os primeiros {num_primes} números primos são: {prime_list}")
```

```python
import math

def newton_sqrt(n):
    if n < 0:
        raise ValueError("Não é possível calcular a raiz quadrada de um número negativo.")
    if n == 0:
        return 0.0

    b = n
    while True:
        p = (b + (n / b)) / 2
        if abs(n - (p * p)) < 0.0001:
            return p
        b = p

number_to_sqrt = float(input("Digite um número para calcular a raiz quadrada: "))

sqrt_newton = newton_sqrt(number_to_sqrt)
print(f"Raiz quadrada (Método de Newton): {sqrt_newton}")

sqrt_math = math.sqrt(number_to_sqrt)
print(f"Raiz quadrada (math.sqrt): {sqrt_math}")
```
